```python
import glob
# 排序
json_paths=glob.glob(r"inference_single_results/*/*_predictions.json")
json_paths.sort()
print(json_paths)
print(len(json_paths))
```

```python
import json
from utils import calculate_bcg_metrics

# 初始化存储各个指标的列表
mean_rrs = []
sdnns = []
rmssds = []

# 遍历所有预测结果JSON文件
for item_path in json_paths:
    # 读取JSON文件
    with open(item_path, 'r') as f:
        data_list = json.load(f)
        
        # 初始化当前文件的累计指标
        mean_rr_sum = 0
        sdnn_sum = 0
        rmssd_sum = 0
        
        # 计算当前文件中所有数据的指标
        for data in data_list:
            peaks = data.get("detected_peaks", [])
            metrics = calculate_bcg_metrics(peaks)
            
            mean_rr_sum += metrics.get("Mean_RR_ms", 0)
            sdnn_sum += metrics.get("SDNN_ms", 0)
            rmssd_sum += metrics.get("RMSSD_ms", 0)
        
        # 计算当前文件的平均值并保存
        num_samples = len(data_list)
        
        mean_rrs.append(round(mean_rr_sum / num_samples, 1))
        sdnns.append(round(sdnn_sum / num_samples, 1))
        rmssds.append(round(rmssd_sum / num_samples, 1))
        print(f"Processed {item_path}: Mean_RR_ms={mean_rrs[-1]}, SDNN_ms={sdnns[-1]}, RMSSD_ms={rmssds[-1]}")

# 计算并输出所有文件的总体平均指标
print(f"Overall Mean_RR_ms: {round(sum(mean_rrs) / len(mean_rrs), 1)}")
print(f"Overall SDNN_ms: {round(sum(sdnns) / len(sdnns), 1)}")
print(f"Overall RMSSD_ms: {round(sum(rmssds) / len(rmssds), 1)}")

# 保存结果到csv文件
import pandas as pd
results_df = pd.DataFrame({
    "File": json_paths,
    "Mean_RR_ms": mean_rrs,
    "SDNN_ms": sdnns,
    "RMSSD_ms": rmssds
})
results_df.to_csv("bcg_rr_metrics_summary.csv", index=False)
```
